```python
import glob
import nbformat
from nbconvert import HTMLExporter

# Find all notebooks in the folder
notebooks = glob.glob("*.ipynb")

if not notebooks:
    print("❌ No notebooks found. Please upload the .ipynb file first.")
else:
    for nb in notebooks:
        print(f"🔄 Converting '{nb}'...")

        # Open and read the notebook
        with open(nb, 'r', encoding='utf-8') as f:
            nb_node = nbformat.read(f, as_version=4)

        # ✨ Delete the corrupted Colab widget metadata ✨
        if "widgets" in nb_node.metadata:
            del nb_node.metadata["widgets"]

        # Convert it to HTML
        html_exporter = HTMLExporter()
        html_exporter.theme = 'light'
        (body, resources) = html_exporter.from_notebook_node(nb_node)

        # ✨ THE ULTIMATE CSS FIX: Obliterate all red backgrounds everywhere ✨
        css_fix = """
<style>
/* 1. Target standard stderr and warnings */
.output_stderr, .output_error, [data-mime-type="application/vnd.jupyter.stderr"] {
    background-color: transparent !important;
    background: transparent !important;
    color: inherit !important;
}

/* 2. Target Jupyter's ANSI color codes (red text and red backgrounds) */
.ansi-red-fg, .ansi-red-intense-fg {
    color: inherit !important;
}
.ansi-red-bg, .ansi-red-intense-bg {
    background-color: transparent !important;
    background: transparent !important;
}

/* 3. Strip background from any preformatted output text boxes just in case */
div.output_subarea pre, div.output_area pre {
    background-color: transparent !important;
    background: transparent !important;
}
</style>
</head>"""
        body = body.replace('</head>', css_fix)

        # Save the new HTML file
        out_name = nb.replace('.ipynb', '.html')
        with open(out_name, 'w', encoding='utf-8') as f:
            f.write(body)

        print(f"✅ Success! Saved as {out_name}. Click 'Refresh' 🔄 in the folder panel to download it.")
```
